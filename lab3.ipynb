```python
# Import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
# Load Dataset

df=pd.read_csv("dataset_2191_sleep.csv")

# Display first 5 rows
df.head()
```

### Check Missing Values

```python
# Check all values
print(df.isna())
```

```python
# count how many missing values
print(df.isna().sum())
```

```python
# checking duplicate rows
df.duplicated()[df.duplicated()==True]
```

```python
# finding number of rows and columns

print("Shape (rows, columns): ", df.shape,"\n")

print("number of rows: ", df.shape[0])
print("number of columns: ", df.shape[1])
```

```python
# viewing the data types of columns
df.dtypes
```

```python
# my data does not contains columns Date and Amount.

df['Date'] = pd.to_datetime(df['Date'])
df['Amount'] = df['Amount'].replace('[\$,]', '', regex=True)
df['Amount'] = pd.to_numeric(df['Amount'])

df.dtypes
```

```python
# Statistical summary (Some statistics using mathematics)
df.describe(include='all')
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['sleep_exposure_index'], bins=20)
plt.title(" Sleep Exposure Levels")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['max_life_span'], bins=20)
plt.title("Maximum Life Span in Mammalse")
plt.show()
```

```python
country_revenue = df.groupby('danger_index')['total_sleep'].sum().sort_values(ascending=False)

plt.figure(figsize=(10,5))
country_revenue.plot(kind='bar')
plt.title("Average Daily Sleep by Danger Index")
plt.ylabel("Revenue")
plt.show()

country_revenue
```

```python
product_revenue = df.groupby('predation_index')['gestation_time'].sum().sort_values(ascending=False)

plt.figure(figsize=(10,5))
product_revenue.plot(kind='bar')
plt.title("Average Gestation Time by Predation Level")
plt.ylabel("Revenue")
plt.show()

product_revenue
```

```python
salesperson_revenue = df.groupby('danger_index')['total_sleep'].sum().sort_values(ascending=False)

salesperson_revenue.head(10)
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='body_weight', y='brain_weight', data=df)
plt.title("Relationship between Body Weight and Brain Weight")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.heatmap(df[['sleep_exposure_index', 'sleep_exposure_index']].corr(), annot=True)
plt.title("Correlation Matrix")
plt.show()
```

```python

sleep_trend = df.groupby('max_life_span')['sleep_exposure_index'].sum()

plt.figure(figsize=(10,5))
sleep_trend.plot()
plt.title("Average Sleep Trend by Exposure Index")
plt.ylabel("Hours of Sleep")
plt.show()
```
