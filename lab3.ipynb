This block reveals the format of the data stored in each column. It shows that variables like body_weight and total_sleep are stored as float64 , while others like danger_index are stored as int64

```python
# my data does not contains columns Date and Amount.

df['Date'] = pd.to_datetime(df['Date'])
df['Amount'] = df['Amount'].replace('[\$,]', '', regex=True)
df['Amount'] = pd.to_numeric(df['Amount'])

df.dtypes
```

This block performs "data cleaning" to fix formatting issues in specific columns:

Date Conversion: It converts the 'Date' column from a simple text string into a formal datetime object so Python can recognize time-based patterns.

Cleaning Currency: It uses a regular expression (regex) to strip away dollar signs ($) and commas (,) from the 'Amount' column.

Numeric Conversion: After removing the symbols, it converts 'Amount' into a numeric format so you can perform math (like sums or averages) on it.

```python
# Statistical summary (Some statistics using mathematics)
df.describe(include='all')
```

This block generates a comprehensive mathematical overview of the entire dataset. It calculates key metrics such as the average, standard deviation , minimum and maximum values, and percentiles for every column.

```python
plt.figure(figsize=(8,5))
sns.histplot(df['sleep_exposure_index'], bins=20)
plt.title(" Sleep Exposure Levels")
plt.show()
```

This block creates a histogram to visualize the distribution of the "sleep_exposure_index" column. It sets a custom window size of 8x5 inches and uses 20 "bins" (vertical bars) to group the data points. The resulting chart allows you to quickly see the frequency of different exposure levels, showing that a level of 1.0 is the most common occurrence in this dataset.

```python
plt.figure(figsize=(8,5))
sns.histplot(df['max_life_span'], bins=20)
plt.title("Maximum Life Span in Mammalse")
plt.show()
```

This block creates a histogram to visualize how the data is spread out. By setting the "bins" to 20, it groups the mammals by their life spans into vertical bars. The chart clearly shows that a life span between 0 and 10 years is the most common in this dataset.

```python
country_revenue = df.groupby('danger_index')['total_sleep'].sum().sort_values(ascending=False)

plt.figure(figsize=(10,5))
country_revenue.plot(kind='bar')
plt.title("Average Daily Sleep by Danger Index")
plt.ylabel("Revenue")
plt.show()

country_revenue
```

This block performs a grouped analysis to see how the danger index affects sleep. It groups the data by danger_index and sums up the total_sleep for each group, sorting the results from highest to lowest. 

```python
product_revenue = df.groupby('predation_index')['gestation_time'].sum().sort_values(ascending=False)

plt.figure(figsize=(10,5))
product_revenue.plot(kind='bar')
plt.title("Average Gestation Time by Predation Level")
plt.ylabel("Revenue")
plt.show()

product_revenue
```

Danger Index: The analysis shows that mammals in the lowest danger category (1) have the highest combined sleep time at 235.5 hours, while those in the highest danger (5) sleep the least at 28.5 hours.



```python
salesperson_revenue = df.groupby('danger_index')['total_sleep'].sum().sort_values(ascending=False)

salesperson_revenue.head(10)
```

These sections use grouping to find totals for specific categories. One analysis shows that mammals in the lowest danger category (1) have the highest combined sleep time at 235.5 hours. The bar chart visualizes this relationship, showing that as the danger index increases, total sleep hours decrease significantly.

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x='body_weight', y='brain_weight', data=df)
plt.title("Relationship between Body Weight and Brain Weight")
plt.show()
```

This block creates a scatter plot to visualize the correlation between body_weight and brain_weight. Each dot on the chart represents a mammal. The distribution shows a clear positive trend: as body weight increases, brain weight 

```python
plt.figure(figsize=(6,4))
sns.heatmap(df[['sleep_exposure_index', 'brain_weight']].corr(), annot=True)
plt.title("Correlation Matrix")
plt.show()
```

The Result (0.37): This number tells us that as sleep exposure increases, brain weight also tends to increase slightly.


```python

sleep_trend = df.groupby('max_life_span')['sleep_exposure_index'].sum()

plt.figure(figsize=(10,5))
sleep_trend.plot()
plt.title("Average Sleep Trend by Exposure Index")
plt.ylabel("Hours of Sleep")
plt.show()
```

This line chart shows the total sleep exposure summed by the maximum life span of the mammals. The peaks and valleys represent how exposure levels fluctuate across different age groups in data.
