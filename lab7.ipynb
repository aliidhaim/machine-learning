 Data Cleaning :


```python
plt.figure(figsize=(12, 7))
sns.boxplot(x='Pclass', y='Age', hue='Pclass', data=train, palette='winter', legend=False)
```

```python
def impute_age(cols):
    Age = cols['Age']
    Pclass = cols['Pclass']

    if pd.isnull(Age):
        if Pclass == 1:
            return 37
        elif Pclass == 2:
            return 29
        else:
            return 24
    else:
        return Age
```

```python
train['Age'] = train[['Age', 'Pclass']].apply(impute_age, axis=1)
```

```python
sns.heatmap(train.isnull(), yticklabels=False, cbar=False, cmap='viridis')
```

```python
train.drop('Cabin', axis=1, inplace=True)
```

```python
train.head()
```

```python
train.dropna(inplace=True)
```

```python
train.head()
```

```python
train.dropna(inplace=True)
```

Converting Categorical Features : 

```python
train.info()
```

```python
sex = pd.get_dummies(train['Sex'], drop_first=True)
embark = pd.get_dummies(train['Embarked'], drop_first=True)
```

drop_first=True removes one category to avoid the dummy variable trap 


```python
train.drop(['Sex', 'Embarked', 'Name', 'Ticket'], axis=1, inplace=True)
```

```python
train = pd.concat([train, sex, embark], axis=1)
```

```python
train.head()
```

Train Test Split : 

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    train.drop('Survived', axis=1),
    train['Survived'],
    test_size=0.30,
    random_state=101)
```

Training and Predicting : 

```python
from sklearn.linear_model import LogisticRegression

logmodel = LogisticRegression(max_iter=1000)
logmodel.fit(X_train, y_train)
```

```python
predictions = logmodel.predict(X_test)
```

Evaluation : 

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, predictions))
```

```python
from sklearn.metrics import confusion_matrix

print(confusion_matrix(y_test, predictions))
```

# Advertising Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
ad_data = pd.read_csv('advertising.csv')
```

```python
ad_data.head()
```

```python
ad_data.info()
```

```python
ad_data.describe()
```

Histogram of Age :

```python
ad_data['Age'].hist(bins=30, color='darkred', alpha=0.7)
plt.xlabel('Age')
plt.title('Distribution of Age')
plt.show()
```

Jointplot - Area Income vs Age :

```python
sns.jointplot(x='Age', y='Area Income', data=ad_data, alpha=0.5)
plt.show()
```

Jointplot: Daily Time on Site vs Age 

```python
sns.jointplot(x='Age', y='Daily Time Spent on Site', data=ad_data, color='red', kind='kde')
plt.show()
```

Jointplot: Daily Time on Site vs Daily Internet Usage : 

```python
sns.jointplot(x='Daily Time Spent on Site', y='Daily Internet Usage', data=ad_data, color='green')
plt.show()
```

Pairplot by Clicked on Ad :

```python
sns.pairplot(ad_data, hue='Clicked on Ad')
plt.show()
```

Train Test Split :

```python
from sklearn.model_selection import train_test_split

X = ad_data[['Daily Time Spent on Site', 'Age', 'Area Income',
             'Daily Internet Usage', 'Male']]
y = ad_data['Clicked on Ad']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=101)
```

Train and Fit the Model : 

```python
from sklearn.linear_model import LogisticRegression

logmodel = LogisticRegression(max_iter=1000)
logmodel.fit(X_train, y_train)
```

 Predictions and Evaluations :

```python
predictions = logmodel.predict(X_test)
```

Classification Report


```python
from sklearn.metrics import classification_report

print(classification_report(y_test, predictions))
```

```python
from sklearn.metrics import confusion_matrix

print(confusion_matrix(y_test, predictions))
```
